# Exploring the Algebra of the 01, 1, 0, 001, 00001000111 cycles

## Original (a,b,c) tuple with Up, Down functions for Shorcut Form:

This was the lattice algebra original identified in the lattice. And it is easiest to see how the period cycle is built in to the positive lattice in this form.

$\tau  = (a, b, c)$  
$V({\tau}) = \frac{2^a - c}{3^b}$  
${\tau}_0  = (0, 0, 0)$  
$f_{up}({\tau})  \mapsto (a+1, b, c)$  
$f_{dn}({\tau})  \mapsto (a+1, b+1, 3c + 2^a)$  

We can see that $f_{up}(f_{dn}({\tau})) \mapsto (a+2, b+1, 3c + 2^a)$

And so we have 
$$
f_{up}(f_{dn}({\tau})) = \frac{2^{a+2} - (3c + 2^a)}{3^{b+1}}
$$
$$
f_{up}(f_{dn}({\tau})) = \frac{4\cdot2^{a} - 3c - 2^a}{3^{b+1}}
$$
$$
f_{up}(f_{dn}({\tau})) = \frac{3\cdot2^{a} - 3c}{3 \cdot 3^{b}}
$$
$$
\frac{2^a - c}{3^b} = V({\tau})
$$



# Mapping $\tau$ tuple to the current:

- $a_{\tau} = a_T$
- $b_{\tau} = -b_T$
- $c_{\tau} = 3^{-b_T}\sum_{(c,d)\in L_T}2^c3^d$
- $f_{up} \mapsto F_1$
- $f_{dn} \mapsto F_0$


## Adding root value to period 2 equations:

$V_{r}({\tau}) = \frac{r2^a - c}{3^b}$  

$$
V_{r}(f_{up}(f_{dn}({\tau}))) = \frac{r2^{a+2} - (3c + 2^a)}{3^{b+1}}
$$
$$
V_{r}(f_{up}(f_{dn}({\tau}))) = \frac{4r\cdot2^{a} - 3c - 2^a}{3^{b+1}}
$$
$$
V_{r}(f_{up}(f_{dn}({\tau}))) = \frac{(4r-1)\cdot2^{a} - 3c}{3 \cdot 3^{b}}
$$
$$
V_{r}(f_{up}(f_{dn}({\tau}))) = \frac{(r-1)\cdot2^{a}}{3^{b+1}} + V_{r}({\tau})
$$

And we see that unless $r=1$, we will not get a cycle with period 2.

# $V_{-1}$, "0" cycle

$V_{-1}({\tau}) = \frac{-2^a - c}{3^b}$  

$f_{dn}({\tau})  \mapsto (a+1, b+1, 3c - 2^a)$ 

$$
V_{-1}(f_{dn}({\tau})) = \frac{-2^{a+1} - 3c - 2^a}{3^{b+1}}
$$
$$
V_{-1}(f_{dn}({\tau})) = \frac{-2 \cdot 2^{a} - 3c - 2^a}{3 \cdot 3^{b}}
$$
$$
V_{-1}(f_{dn}({\tau})) = \frac{-3 \cdot 2^{a} - 3c}{3 \cdot 3^{b}}
$$
$$
V_{-1}(f_{dn}({\tau})) = \frac{-2^{a} - c}{3^{b}}
$$
$$
V_{-1}(f_{dn}({\tau})) = V_{-1}({\tau})
$$


## Adding root value to $V_{-1}$, "0" cycle equations:

$$
V_{r}(f_{dn}({\tau})) = \frac{r \cdot 2^{a+1} - 3c - 2^a}{3^{b+1}}
$$
$$
V_{r}(f_{dn}({\tau})) = \frac{r \cdot 2 \cdot 2^{a} - 3c - 2^a}{3 \cdot 3^{b}}
$$
$$
V_{r}(f_{dn}({\tau})) = \frac{((r+1) -3\cdot 2)2^{a} - 3c}{3^{b+1}}
$$
$$
V_{r}(f_{dn}({\tau})) = \frac{r+1}{3^{b+1}} +  V_{-1}({\tau})
$$

And we see we can only have a cycle when $r = -1$

# $V_{-5}$, "001" cycle

$f_{up}(f_{dn}(f_{dn}({\tau}))) \mapsto (a+3, b+2, 3(3c + 2^a) + 2^{a+1})$  
$\mapsto (a+3, b+2, 9c + 5 \cdot 2^a )$


$V_{-5}({\tau}) = \frac{-5 \cdot 2^a - c}{3^b}$  

$$
V_{-5}(f_{up}(f_{dn}(f_{dn}({\tau})))) = \frac{-5 \cdot 2^{a+3} - (9c + 5 \cdot 2^{a})}{3^{b+2}}
$$
$$
V_{-5}(f_{up}(f_{dn}(f_{dn}({\tau})))) = \frac{-9\cdot 5 \cdot 2^{a} - 9c }{9 \cdot 3^{b}}
$$
$$
\frac{-5 \cdot 2^a - c}{3^b} = V_{-5}({\tau})
$$
