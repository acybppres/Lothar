# Positive Lattice Divergence

Lets look at how the three parts interact to give us diverging lattices for $r > 2$


$$
\begin{aligned}
	V_{3k+j}(\sigma)
	&= \underbrace{((3k)+1)V_{0}(\sigma) - (3k)V_{-1}(\sigma)}_{\text{dominant expanding}}
	\\[6pt]
	&\quad + j \cdot \underbrace{(2V_{0}(\sigma) - V_{-1}(\sigma))}_{\text{period 2}}
	\\[6pt]
	&\quad - j \cdot \underbrace{V_{0}(\sigma)}_{\text{period 1}}, 
	\qquad j \in (1,2)
\end{aligned}
$$

We used the "0...01...1" label as a consitent divergence metric across lattices.  This label should be approximately non-divergent in a lattice that has a cycle of the expected length for the given root value, but we see for almost all cases $r > 2$ it is instead strongly divergent and that as $k$ increases the trend is for greater divergence.

Tweaked the pattern to start with "1" for $3k+2$ case to get non-divergent 2.

```python
a_over_b = math.log(2, 3)
```

```python
# 3k+1 cases 
#
for r in range(1, 70, 3):
    len_cycle = round(round(abs(r) * a_over_b)+1)
    zero_count = round(len_cycle * a_over_b)
    one_count = len_cycle - zero_count
    print((r, zero_count, one_count))
    label = ("0" * zero_count) + ("1" * one_count)
    T = mrTupFromPath(label)
    val = mrTupRootedValue(r, T)
    label_ = label+label+label
    T_ = mrTupFromPath(label_)
    val_ = mrTupRootedValue(r, T_)
    label__ = label_+label_+label_
    T__ = mrTupFromPath(label__)
    val__ = mrTupRootedValue(r, T__)

    divergence = (val[0]/val[1]) - (val_[0]/val_[1])
    divergence_ = (val_[0]/val_[1]) - (val__[0]/val__[1])
    
    print((r, divergence, divergence_))
#
```

```python
# 3k+2 cases 
#
for r in range(2, 70, 3):
    len_cycle = round(round(abs(r) * a_over_b)+1)
    zero_count = round(len_cycle * a_over_b)
    one_count = len_cycle - zero_count
    print((r, zero_count, one_count))
    label = "1" + ("0" * zero_count) + ("1" * (one_count - 1))
    T = mrTupFromPath(label)
    val = mrTupRootedValue(r, T)
    label_ = label+label+label
    T_ = mrTupFromPath(label_)
    val_ = mrTupRootedValue(r, T_)
    label__ = label_+label_+label_
    T__ = mrTupFromPath(label__)
    val__ = mrTupRootedValue(r, T__)

    divergence = (val[0]/val[1]) - (val_[0]/val_[1])
    divergence_ = (val_[0]/val_[1]) - (val__[0]/val__[1])
    
    print((r, divergence, divergence_))
#
```
